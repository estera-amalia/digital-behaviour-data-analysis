```python
import pandas as pd #reading the data set
df=pd.read_csv("C:\\mental_health_digital_behavior_data.csv")
print(df)
```

```python
print(df.head(10)) #printing first 10 rows
print(df.tail(10)) #printing last 10 rows
print(df.info()) #printing information about the dataset
```

```python
x=df.dropna() #removing empty cells
print(x)
```

```python
x=df.drop_duplicates() #drop duplicate data
print(x)
```

```python
x=df.loc[2,"digital_wellbeing_score"]
print(x)
```

```python
import numpy
x=df["anxiety_level"].mean() #calculating the average of a chosen column
print(x)
df["anxiety_level"].fillna(x, inplace=True)
print(df.head(3))
```

```python
x=df["mood_score"].median() #calculating the middle value
print(x)
df["mood_score"].fillna(x, inplace=True)
print(df.head(3))
```

```python
x=df["focus_score"].mode() #most frequent value
print(x)
df["focus_score"].fillna(x, inplace=True)
print(df.head(3))
```

```python
x=df.loc[2,"digital_wellbeing_score"]=52.3 #change specific wrong data from row
print(x)
```
